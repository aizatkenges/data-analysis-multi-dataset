```python
import pandas as pd

df = pd.read_csv("netflix_titles.csv")  

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
# Check for missing values
df.isnull().sum()
```

```python
df['director'].fillna('Unknown', inplace=True)
```

```python
df['cast'].fillna('Unknown', inplace=True)
```

```python
df['country'].fillna('Unknown', inplace=True)
```

```python
df.dropna(subset=['date_added', 'duration', 'rating'], inplace=True)
```

```python
# handling outliers
import seaborn as sns
import matplotlib.pyplot as plt

# Boxplot to visualize outliers
plt.figure(figsize=(10, 5))
sns.boxplot(x=df['release_year'])
plt.title("Boxplot of Release Year")
plt.show()
```

```python
Q1 = df['release_year'].quantile(0.25)  # 25th percentile
Q3 = df['release_year'].quantile(0.75)  # 75th percentile
IQR = Q3 - Q1  # Interquartile range

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['release_year'] < lower_bound) | (df['release_year'] > upper_bound)]
print(outliers)
```

```python
print(sorted(outliers['release_year'].unique()))
```

```python
#data preparation

df["duration"].dtype
```

```python
#change to numeric values
df["duration"] = df["duration"].str.extract("(\d+)").astype(float)
```

```python
df["duration"]
```

```python
df["decade"] = (df["release_year"] // 10) * 10
```

```python
df["decade"]
```

```python
#data analysis
movies_per_decade = df["decade"].value_counts().sort_index()
print(movies_per_decade)
```

```python
# Get basic statistics for movie duration
print(df["duration"].describe())

# Check the distribution of duration
print(df["duration"].value_counts().sort_index())
```

```python
# Remove movies with duration less than 30 minutes
df = df[df["duration"] >= 30]

print(df["duration"].describe())
```

```python
# Find the most common durations
print(df["duration"].value_counts().head(10))
```

```python


#popular movies for decades
print(df["decade"].value_counts().sort_index())
```

```python
# Count of movies per rating category
print(df["rating"].value_counts())
```

```python
# Calculate average duration per rating
print(df.groupby("rating")["duration"].mean().sort_values(ascending=False))
```

```python
# Calculate average duration per decade
print(df.groupby("decade")["duration"].mean().sort_index())
```

```python
# Check the shortest and longest movies
print(df["duration"].min(), df["duration"].max())
```

```python
#Recommendation with plots
# What duration should they focus on?
# Find the most common movie durations
import matplotlib.pyplot as plt

df["duration"].hist(bins=10)  
plt.xlabel("Duration (minutes)")
plt.ylabel("Number of Movies")
plt.title("Distribution of Movie Durations")
plt.show()
```

```python
# What rating category is most popular?
# Count movies per rating
df["rating"].value_counts().plot(kind="bar")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.title("Number of Movies per Rating")
plt.show()
```

```python
#What decade had the most high-rated films?
# Find the most popular decades
df.groupby("decade")["rating"].count().plot(kind="bar")
plt.xlabel("Decade")
plt.ylabel("Number of Movies")
plt.title("Number of Movies Released per Decade")
plt.show()
```

```python
#visualization


import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
sns.histplot(df["duration"], bins=50, kde=True)
plt.xlabel("Duration (minutes)")
plt.ylabel("Count of Movies")
plt.title("Distribution of Movie Durations")
plt.show()
```

```python
#duration per rating
plt.figure(figsize=(12, 6))
sns.boxplot(x="rating", y="duration", data=df)
plt.xlabel("Movie Rating")
plt.ylabel("Duration (minutes)")
plt.title("Movie Duration by Rating")
plt.xticks(rotation=45)
plt.show()
```

```python
#Count of Movies per Decade

df["decade"] = (df["release_year"] // 10) * 10  # Create decade column

plt.figure(figsize=(12, 6))
sns.countplot(x=df["decade"], palette="viridis")
plt.xlabel("Decade")
plt.ylabel("Number of Movies")
plt.title("Number of Movies Released per Decade")
plt.xticks(rotation=45)
plt.show()
```
