```python
import pandas as pd

df = pd.read_excel("Online Retail.xlsx")  # Load Excel file
df.head()  # Display first 5 rows
```

```python
df.info()
```

```python
# Summary statistics for numerical variables
df.describe().T
```

```python
# Summary statistics for categorical variables
df.describe(include='object').T
```

```python
 df.isnull().sum()
```

```python
# Fill missing Description and Customer id  with "Unknown Product"
df["Description"].fillna("Unknown Product", inplace=True)

df["CustomerID"].fillna("Unknown", inplace=True)
```

```python
print(df.isnull().sum())
```

```python
df.drop_duplicates()  # Remove duplicates
```

```python
df["TotalPrice"] = df["Quantity"] * df["UnitPrice"]  # Calculate total spend per transaction
```

```python
# 🔹 Metric 1: Total Revenue
total_revenue = df["TotalPrice"].sum()
```

```python
# 🔹 Metric 2: Average Revenue per Transaction
avg_revenue_per_txn = df.groupby("InvoiceNo")["TotalPrice"].sum().mean()
```

```python
# 🔹 Metric 3: Number of Unique Customers
unique_customers = df["CustomerID"].nunique()
```

```python
# 🔹 Metric 4: Customer Lifetime Value (LTV)
ltv = df.groupby("CustomerID")["TotalPrice"].sum().reset_index()
ltv.columns = ["CustomerID", "LTV"]
```

```python
# 🔹 Metric 5: Retention Rate
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# 🔹 Metric 5: Retention Rate
first_purchase = df.groupby("CustomerID")["InvoiceDate"].min().reset_index()
first_purchase.columns = ["CustomerID", "FirstPurchaseDate"]
df = df.merge(first_purchase, on="CustomerID")
df["IsReturning"] = df["InvoiceDate"] > df["FirstPurchaseDate"]
retention_rate = df["IsReturning"].mean()
```

```python
# 🔹 Metric 6: Churn Rate (Customers who didn’t return after 90 days)
latest_date = df["InvoiceDate"].max()
df["DaysSinceLastPurchase"] = (latest_date - df["InvoiceDate"]).dt.days
churn_threshold = 90
churn_rate = (df["DaysSinceLastPurchase"] > churn_threshold).mean()
```

```python
# 🔹 Metric 7: Average Order Value (AOV)
aov = df.groupby("InvoiceNo")["TotalPrice"].sum().mean()
```

```python
# 🔹 Metric 8: Total Number of Transactions
total_transactions = df["InvoiceNo"].nunique()
```

```python
# 🔹 Metric 9: Average Items per Transaction
avg_items_per_txn = df.groupby("InvoiceNo")["Quantity"].sum().mean()

# 🔹 Metric 10: Most Sold Product
most_sold_product = df.groupby("Description")["Quantity"].sum().idxmax()

# 🔹 Metric 11: Country with the Most Orders
top_country = df["Country"].value_counts().idxmax()

# 🔹 Metric 12: Customer with the Highest Spending
top_customer = ltv.sort_values(by="LTV", ascending=False).iloc[0]["CustomerID"]

# 🔹 Metric 13: Best Month, Day, Hour for Sales
df["Month"] = df["InvoiceDate"].dt.month
df["DayOfWeek"] = df["InvoiceDate"].dt.dayofweek
df["Hour"] = df["InvoiceDate"].dt.hour

best_month = df.groupby("Month")["TotalPrice"].sum().idxmax()
best_day = df.groupby("DayOfWeek")["TotalPrice"].sum().idxmax()
best_hour = df.groupby("Hour")["TotalPrice"].sum().idxmax()
```

```python
# 🔹 Metric 14: Revenue Distribution by Country
revenue_by_country = df.groupby("Country")["TotalPrice"].sum().sort_values(ascending=False)

# 🔹 Metric 15: Repeat Customer Rate
repeat_customers = df.groupby("CustomerID")["InvoiceNo"].nunique().gt(1).mean()
```

```python
# 🔹 Visualization 1: Customer LTV Distribution
plt.figure(figsize=(10, 5))
sns.histplot(ltv["LTV"], bins=50, kde=True)
plt.title("Customer Lifetime Value (LTV) Distribution")
plt.xlabel("LTV")
plt.ylabel("Number of Customers")
plt.show()
```

```python
# 🔹 Visualization 2: Monthly Sales Trend
monthly_sales = df.groupby("Month")["TotalPrice"].sum()
plt.figure(figsize=(10, 5))
monthly_sales.plot(kind="bar", color="skyblue")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

```python
# 🔹 Visualization 3: Revenue by Country
plt.figure(figsize=(12, 5))
revenue_by_country.head(10).plot(kind="bar", color="orange")
plt.title("Top 10 Countries by Revenue")
plt.xlabel("Country")
plt.ylabel("Total Revenue")
plt.show()
```

```python
# 🔹 Print Key Metrics
print(f"🔹 Total Revenue: £{total_revenue:,.2f}")
print(f"🔹 Average Revenue per Transaction: £{avg_revenue_per_txn:,.2f}")
print(f"🔹 Number of Unique Customers: {unique_customers}")
print(f"🔹 Average Customer LTV: £{ltv['LTV'].mean():,.2f}")
print(f"🔹 Retention Rate: {retention_rate * 100:.2f}%")
print(f"🔹 Churn Rate: {churn_rate * 100:.2f}%")
print(f"🔹 Average Order Value (AOV): £{aov:,.2f}")
print(f"🔹 Total Transactions: {total_transactions}")
print(f"🔹 Average Items per Transaction: {avg_items_per_txn:.2f}")
print(f"🔹 Most Sold Product: {most_sold_product}")
print(f"🔹 Country with the Most Orders: {top_country}")
print(f"🔹 Customer with the Highest Spending: {top_customer}")
print(f"🔹 Best Month for Sales: {best_month}")
print(f"🔹 Best Day for Sales: {best_day}")
print(f"🔹 Best Hour for Sales: {best_hour}")
print(f"🔹 Repeat Customer Rate: {repeat_customers * 100:.2f}%")
```
