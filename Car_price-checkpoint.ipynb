```python
import pandas as pd 
cols = ['msrp', 'askPrice', 'mileage', 'isNew', 'color', 'interiorColor',
        'vf_VehicleType', 'vf_WheelBaseLong', 'vf_WheelSizeFront', 'vf_Windows']

df = pd.read_csv("CIS_Automotive_Kaggle_Sample.csv", usecols=cols, nrows=7000)
df.head()
```

```python
df.info()
```

```python
print(df.dtypes)
```

```python
df.drop_duplicates(inplace=True)
```

```python
missing = df.isnull().sum()
print("Пропущенные значения:\n", missing)
```

```python
# 1. color и interiorColor — заполняем 'Unknown'
df['color'] = df['color'].fillna('Unknown')
df['interiorColor'] = df['interiorColor'].fillna('Unknown')
```

```python
# 2. vf_VehicleType — удалим 1 строку с NaN
df = df.dropna(subset=['vf_VehicleType'])
```

```python
# 3. vf_WheelBaseLong — слишком много NaN (60%)
df = df.drop(columns=['vf_WheelBaseLong'])
```

```python
# 4. vf_WheelSizeFront — заполняем медианой
df['vf_WheelSizeFront'] = df['vf_WheelSizeFront'].fillna(df['vf_WheelSizeFront'].median())
```

```python
# 5. vf_Windows — много пропусков, заменим на 'Unknown'
df['vf_Windows'] = df['vf_Windows'].fillna('Unknown')
```

```python
# Проверим снова
print("После обработки:\n", df.isnull().sum())
```

```python
#Handling Outliers
import seaborn as sns
import matplotlib.pyplot as plt

# Boxplot для выявления выбросов
sns.boxplot(x=df['msrp'])
plt.title('MSRP Outliers')
plt.show()
```

```python
# Удалить или ограничить выбросы
q1 = df['msrp'].quantile(0.25)
q3 = df['msrp'].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

df = df[(df['msrp'] >= lower) & (df['msrp'] <= upper)]
```

```python
#Data Preparation
from sklearn.preprocessing import MinMaxScaler, LabelEncoder

# Масштабирование числовых данных
scaler = MinMaxScaler()
df[['msrp_scaled', 'askPrice_scaled', 'mileage_scaled']] = scaler.fit_transform(df[['msrp', 'askPrice', 'mileage']])

# Преобразование категориальных
le = LabelEncoder()
df['color_encoded'] = le.fit_transform(df['color'])
df['vehicleType_encoded'] = le.fit_transform(df['vf_VehicleType'].astype(str))

# Новая переменная: разница между ценами
df['price_diff'] = df['msrp'] - df['askPrice']

```

```python
# Data Analysis with Pandas
# Описательные статистики
print(df[['msrp', 'askPrice', 'mileage']].describe())

# Корреляции
print(df[['msrp', 'askPrice', 'mileage']].corr())

# Частоты категорий
print(df['color'].value_counts())
```
